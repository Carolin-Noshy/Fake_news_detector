```python
df.columns=df.columns.str.strip().str.lower().str.replace(" ",'_')
```

```python
df["id"] = pd.to_numeric( df["id"].astype(str).str.rstrip(".json"),errors="coerce")
```

```python
text = ["label", "statement", "subject", "speaker", "job", "state", "party", "context"]

for col in text:
    df[col] = df[col].astype("string").str.strip().str.title()
```

```python
df
```

```python
numerical = ["barely_true", "false", "half_true", "mostly_true", "pants_on_fire"]

for cols in numerical:
    df[cols] = pd.to_numeric(df[cols], errors="coerce")
```

```python
unknown_cols=["job","state","context"]
for col in unknown_cols:
  df[col]=df[col].fillna("unknown")
```

```python
median_cols=["barely_true","false","half_true","mostly_true","pants_on_fire"]
for col in median_cols:
  df[col]=df[col].fillna(df[col].median())
```

```python
df[col]=df[col].replace(" ",pd.NA)
```

```python
numerical_cols=["barely_true","false","half_true","mostly_true","pants_on_fire"]
for col in numerical_cols:
  q3=df[col].quantile(75/100)#.075
  q1=df[col].quantile(25/100)
  IQR=q3-q1
  print(col)
  print("Q1 =",q1)
  print("Q3 =",q3)
  print("IQR =",IQR)
```

```python
for col in numerical_cols:
  q1=df[col].quantile(75/100)
  q3=df[col].quantile(25/100)
  IQR=q3-q1
  upper=q3+1.5*IQR
  lower=q3-1.5*IQR
  outliers=df[(df[col]>upper) | (df[col]<lower)]
  print(outliers)
```

```python
mode_cols =[ "barely_true","false","half_true","mostly_true","pants_on_fire"]
for col in mode_cols:
  mode_value=df[col].mode()[0]
  print(col,"mode=",mode_value)
```

```python
mean_cols=["barely_true","false","half_true","mostly_true","pants_on_fire"]
for col in mean_cols:
  df[col]=df[col].fillna(df[col].mean())
```

```python
df["label"] = df["label"].str.strip().str.lower()

counts = df["label"].value_counts()

labels = [
    "False",
    "True"
]

data = [
    counts.get("false", 0),
    counts.get("true", 0)
]

plt.pie(data , labels=labels , autopct="%1.1f%%",explode=(0.1, .2))

plt.title("Truthfulness Distribution")

plt.show()
```
